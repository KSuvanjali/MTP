```python
# STEP 1: Upload file in Colab
from google.colab import files
uploaded = files.upload()
# STEP 2: All analyses with corrected plots and value ranges
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

# Load file
df = pd.read_excel("ALL Seasonal zones.xlsx", skiprows=3)

# Rename columns
df = df.rename(columns={
    'Unnamed: 2': 'Zone',
    'Unnamed: 3': 'Location',
    'Unnamed: 4': 'Season',
    'Unnamed: 5': 'Inhalable',
    'Unnamed: 6': 'Thoracic',
    'Unnamed: 7': 'Alveolic'
})

# Clean data
df.dropna(subset=['Zone', 'Location', 'Season', 'Inhalable', 'Thoracic', 'Alveolic'], inplace=True)
df[['Inhalable', 'Thoracic', 'Alveolic']] = df[['Inhalable', 'Thoracic', 'Alveolic']].apply(pd.to_numeric, errors='coerce')

# Save cleaned version
df.to_excel("Cleaned_Data.xlsx", index=False)
files.download("Cleaned_Data.xlsx")

# Descriptive stats
zone_stats = df.groupby("Zone")[["Inhalable", "Thoracic", "Alveolic"]].describe()
zone_stats.to_excel("zone_descriptive_stats.xlsx")
files.download("zone_descriptive_stats.xlsx")

# Correlation heatmap
corr = df[['Inhalable', 'Thoracic', 'Alveolic']].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix of Particles (µg/m³)")
plt.show()

# Boxplots by season (actual value range shown)
for col in ['Inhalable', 'Thoracic', 'Alveolic']:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x='Season', y=col, data=df)
    plt.title(f'{col} Concentration by Season')
    plt.ylabel(f'{col} (µg/m³)')
    plt.xlabel("Season")
    plt.tight_layout()
    plt.show()

# Zone-wise average
zone_avg = df.groupby("Zone")[["Inhalable", "Thoracic", "Alveolic"]].mean().reset_index()
zone_avg.plot(x="Zone", kind="bar", figsize=(10, 6))
plt.title("Average Particle Concentration by Zone")
plt.ylabel("Concentration (µg/m³)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Location-wise bar plot
location_avg = df.groupby("Location")[["Inhalable", "Thoracic", "Alveolic"]].mean().sort_values("Inhalable", ascending=False)
location_avg.plot(kind='bar', figsize=(12, 6))
plt.title("Average Particle Concentration by Location")
plt.ylabel("Concentration (µg/m³)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# Raw-value scatter matrix (NOT scaled)
sns.pairplot(df[["Inhalable", "Thoracic", "Alveolic"]])
plt.suptitle("Scatter Plot Matrix of Particles (µg/m³)", y=1.02)
plt.show()

# Heatmap: Location × Season for Inhalable
pivot_table = df.pivot_table(values='Inhalable', index='Location', columns='Season', aggfunc='mean')
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu')
plt.title("Average Inhalable (µg/m³) by Location and Season")
plt.xlabel("Season")
plt.ylabel("Location")
plt.tight_layout()
plt.show()

# PCA (standardized only for PCA step)
from sklearn.preprocessing import StandardScaler
X = df[['Inhalable', 'Thoracic', 'Alveolic']]
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

df['PC1'], df['PC2'] = pca_result[:, 0], pca_result[:, 1]

# PCA plot
sns.scatterplot(x='PC1', y='PC2', hue='Zone', data=df)
plt.title("PCA of Particle Concentrations by Zone")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.tight_layout()
plt.show()

# Clustering
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
sns.scatterplot(x='PC1', y='PC2', hue='Cluster', palette='Set2', data=df)
plt.title("K-Means Clustering of Pollution Profiles")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.tight_layout()
plt.show()

# Regression models
model1 = smf.ols("Inhalable ~ C(Zone) + C(Season)", data=df).fit()
print(model1.summary())

model2 = smf.ols("Alveolic ~ Inhalable + Thoracic", data=df).fit()
print(model2.summary())
```
